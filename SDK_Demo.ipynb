```python
# 導入我們剛剛建立的 SDK
from pathlib import Path
from antenna_sdk import CSVParser, PatternAnalyzer
from antenna_sdk.exporters.pptx import PPTXExporter

# 導入用於在 Notebook 中直接繪圖驗證的套件
import matplotlib.pyplot as plt
import numpy as np

# 設定中文字體與圖表樣式
plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False

print("✅ 模組導入成功！")
```

```python
# 指定測試檔案路徑 (請確保 test.csv 在專案根目錄)
csv_file_path = Path("test.csv")

print(f"🔍 開始解析: {csv_file_path.name}")

# 執行 Parser
parser = CSVParser(csv_file_path)
parsed_data = parser.parse()

# 展示解析結果摘要
print("\n📊 解析結果摘要:")
print(f"  - 偵測編碼: {parsed_data.encoding}")
print(f"  - 提取區塊: {list(parsed_data.blocks.keys())}")

# 取得第一個頻率的字串作為示範
first_freq = next(iter(parsed_data.block_frequencies.get("ePhi", [])))[1]
print(f"  - 成功提取的頻率數量: {len(parsed_data.block_frequencies.get('ePhi', []))} 個")
print(f"  - 示範頻率: {first_freq} Hz")
```

```python
print("⚙️ 開始執行核心分析 (矩陣提取與 2D 切面計算)...")

# 執行 Analyzer
analyzer = PatternAnalyzer()
measurement = analyzer.analyze(parsed_data)

# 驗證數據結構
print("\n✅ 分析完成！數據結構驗證:")
print(f"  - ePhi 矩陣數量: {len(measurement.matrices.get('ePhi', {}))}")
print(f"  - 成功計算 2D 切面的頻率數量: {len(measurement.cut_planes)}")

# 檢查單一頻率的切面數據形狀
if measurement.cut_planes:
    demo_freq = list(measurement.cut_planes.keys())[0]
    xy_h_shape = measurement.cut_planes[demo_freq].H["XY"].shape
    print(f"  - 示範頻率 {demo_freq} Hz 的 XY 平面 H 極化數據形狀: {xy_h_shape} (應為 25 點)")
```

```python
# 挑選第一個頻率進行繪圖驗證
if measurement.cut_planes:
    target_freq = list(measurement.cut_planes.keys())[0]
    cp = measurement.cut_planes[target_freq]
    
    # 準備 25 點的角度陣列 (0 到 360 度)
    angles = np.linspace(0, 2 * np.pi, 25, endpoint=True)
    
    # 建立 1x2 的子圖 (H 極化 vs V 極化 的 XY 平面)
    fig, (ax1, ax2) = plt.subplots(1, 2, subplot_kw={'projection': 'polar'}, figsize=(12, 5))
    fig.suptitle(f'即時驗證: {target_freq} Hz - XY Plane (Theta=90°)', fontsize=16, fontweight='bold')
    
    # 繪製 H 極化 (EPHI)
    ax1.plot(angles, cp.H["XY"], color='blue', linewidth=2, label='H (EPHI)')
    ax1.set_title('H Polarization')
    ax1.set_theta_zero_location('N')
    ax1.set_theta_direction(-1)
    ax1.grid(True, linestyle='--', alpha=0.7)
    
    # 繪製 V 極化 (ETHETA)
    ax2.plot(angles, cp.V["XY"], color='red', linewidth=2, label='V (ETHETA)')
    ax2.set_title('V Polarization')
    ax2.set_theta_zero_location('N')
    ax2.set_theta_direction(-1)
    ax2.grid(True, linestyle='--', alpha=0.7)
    
    plt.tight_layout()
    plt.show()
else:
    print("⚠️ 沒有切面數據可供繪圖。")
```

```python
output_ppt_path = Path("notebook_demo_report.pptx")

print(f"🚀 開始匯出 PPT 報告至: {output_ppt_path.name}")

# 呼叫我們之前寫好的 PPTXExporter
exporter = PPTXExporter(dpi=150)
exporter.export(measurement, output_ppt_path)

print(f"🎉 完成！請在專案資料夾中查看 {output_ppt_path.name}")
```
